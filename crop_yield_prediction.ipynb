```python
# Importing required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings

from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from scipy.stats import pearsonr
from sklearn.model_selection import train_test_split, cross_val_score, cross_val_predict, GridSearchCV
from sklearn.preprocessing import PolynomialFeatures, StandardScaler
from sklearn.ensemble import GradientBoostingRegressor
from sklearn import metrics

warnings.filterwarnings('ignore')  # To ignore warnings

# Load dataset
df_cotton = pd.read_csv('../data/dataset.csv')

# Fill missing values with group mean based on 'State'
df_cotton = df_cotton.fillna(df_cotton.groupby('State').transform('mean'))

# Color mapping for visualization
colors = {'Alabama': '#E50800',
          'Arizona': '#D50713',
          'Arkansas': '#C60626',
          'California': '#597AA1',
          'Georgia': '#A7044C',
          'Louisiana': '#98045F',
          'Mississippi': '#880372',
          'Missouri': '#790285',
          'New Mexico': '#690198',
          'North Carolina': '#5A00AB',
          'Oklahoma': '#4B00BF',
          'South Carolina': '#8FD900',
          'Tennessee': '#2FD500',
          'Texas': '#00B9CA'}

# Scatter plot of yield over years colored by state
fig, ax = plt.subplots()
ax.scatter(df_cotton['Year'], df_cotton['Lint Yield (Pounds/Harvested Acre)'],
           c=df_cotton['State'].apply(lambda x: colors[x]))
plt.title("Year vs Lint Yield")
plt.xlabel("Year")
plt.ylabel("Lint Yield (Pounds/Harvested Acre)")
plt.show()

# Histograms for fertilizer application
plt.figure(figsize=(12, 10))
plt.subplot(231)
sns.histplot(df_cotton['Nitrogen (%)'], color='r', bins=100)
plt.title('Nitrogen Area %')

plt.subplot(232)
sns.histplot(df_cotton['Phosphorous (%)'], color='b', bins=100)
plt.title('Phosphorous Area %')

plt.subplot(233)
sns.histplot(df_cotton['Potash (%)'], color='g', bins=100)
plt.title('Potash Area %')

plt.subplot(234)
sns.histplot(df_cotton['Nitrogen (Pounds/Acre)'], color='r', bins=100)
plt.title('Nitrogen (Pounds/Acre)')

plt.subplot(235)
sns.histplot(df_cotton['Phosphorous (Pounds/Acre)'], color='b', bins=100)
plt.title('Phosphorous (Pounds/Acre)')

plt.subplot(236)
sns.histplot(df_cotton['Potash (Pounds/Acre)'], color='g', bins=100)
plt.title('Potash (Pounds/Acre)')
plt.tight_layout()
plt.show()

# Bar plot of total lint yield by state
plt.figure(figsize=(22, 6))
GraphData = df_cotton.groupby(['State'])['Lint Yield (Pounds/Harvested Acre)'].sum().nlargest(10)
GraphData.plot(kind='bar')
plt.ylabel('Lint Yield')
plt.xlabel('State Name')
plt.title('Top 10 States by Total Lint Yield')
plt.show()

# Mapping state names to numbers
mapping = {'Alabama': 1,
           'Arizona': 2,
           'Arkansas': 3,
           'California': 4,
           'Georgia': 5,
           'Louisiana': 6,
           'Mississippi': 7,
           'Missouri': 8,
           'New Mexico': 9,
           'North Carolina': 10,
           'Oklahoma': 11,
           'South Carolina': 12,
           'Tennessee': 13,
           'Texas': 14}
df_cotton = df_cotton.replace({'State': mapping})

# Split data into features and target
x = df_cotton.drop('Lint Yield (Pounds/Harvested Acre)', axis=1)
y = df_cotton['Lint Yield (Pounds/Harvested Acre)']

# Pearson correlation with target
print("\nPearson Correlations:")
for col in x.columns:
    cor, _ = pearsonr(df_cotton[col], y)
    print(f'Correlation between {col} and Lint Yield: {cor:.3f}')

# Train-test split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
print(f"\nTrain shape: {x_train.shape}, Test shape: {x_test.shape}")

# Polynomial features
poly = PolynomialFeatures(degree=2, include_bias=False)
x_train_poly = poly.fit_transform(x_train)
x_test_poly = poly.transform(x_test)

# Feature scaling
scaler = StandardScaler()
x_train_poly = scaler.fit_transform(x_train_poly)
x_test_poly = scaler.transform(x_test_poly)

# Gradient Boosting with GridSearchCV
param_grid = {
    'n_estimators': [300, 500],
    'max_depth': [4, 5, 6],
    'min_samples_split': [2, 5, 10],
    'learning_rate': [0.01, 0.03, 0.05]
}

gbr = GradientBoostingRegressor(loss='squared_error', random_state=42)
grid_search = GridSearchCV(estimator=gbr, param_grid=param_grid, cv=3, n_jobs=-1, scoring='r2')
grid_search.fit(x_train_poly, y_train)

print("\nBest parameters:", grid_search.best_params_)
print("Best CV R2 score:", grid_search.best_score_)

# Fit best model
yield_predict = grid_search.best_estimator_
yield_predict.fit(x_train_poly, y_train)

# Predictions
yield_predict_train = yield_predict.predict(x_train_poly)
yield_predict_test = yield_predict.predict(x_test_poly)

print("\nSample Predictions:")
print(pd.DataFrame({'Actual': y_test.values[:5], 'Predicted': yield_predict_test[:5]}))

# Cross-validation on training set
scores = cross_val_score(yield_predict, x_train_poly, y_train, cv=5)
print("\nCross-validation scores (train):", scores)

predictions = cross_val_predict(yield_predict, x_train_poly, y_train, cv=5)
accuracy = r2_score(y_train, predictions)
print("Cross-validated R2 accuracy (train):", accuracy)

# Plot predictions vs actual
x_ax = range(len(y_test))
plt.figure(figsize=(20, 6))
plt.scatter(x_ax, y_test, s=5, color="blue", label="Actual")
plt.plot(x_ax, yield_predict_test, lw=0.8, color="red", label="Predicted")
plt.legend()
plt.title("Test Set Predictions vs Actual")
plt.show()

# Evaluation metrics
print('\nEvaluation Metrics:')
print('MAE =', mean_absolute_error(y_test, yield_predict_test))
print('MSE =', mean_squared_error(y_test, yield_predict_test))
print('R2 =', r2_score(y_test, yield_predict_test))

# Adjusted R2
n = x_test_poly.shape[0]
k = x_test_poly.shape[1]
adjusted_r2 = 1 - (1 - r2_score(y_test, yield_predict_test)) * ((n - 1) / (n - k - 1))
print('Adjusted R2 =', adjusted_r2)

print('RMSE (train) =', np.sqrt(mean_squared_error(y_train, yield_predict_train)))
print('RMSE (test) =', np.sqrt(mean_squared_error(y_test, yield_predict_test)))

# Yield range
print('Yield range =', df_cotton['Lint Yield (Pounds/Harvested Acre)'].max() - df_cotton['Lint Yield (Pounds/Harvested Acre)'].min())
```
